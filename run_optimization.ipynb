```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "run_optimization.py").is_file():
    raise RuntimeError("Open this notebook from the publication repository root.")
print("Interpreter:", sys.executable)
```

## Campaign settings

Use a unique campaign ID for new production. Configuration values are read from
the INI file. Production requires both an explicit mode and confirmation token.


```python
MODE = "dry-run"
CONFIG_PATH = "config/config.ini"
CAMPAIGN_ID = None
CONFIRM_PRODUCTION = ""
```

```python
if MODE not in {"dry-run", "optimization"}:
    raise ValueError("MODE must be dry-run or optimization.")
arguments = [sys.executable, "run_optimization.py", "--config", CONFIG_PATH]
if CAMPAIGN_ID:
    arguments.extend(["--campaign-id", CAMPAIGN_ID])
if MODE == "dry-run":
    arguments.append("--dry-run")
elif not CAMPAIGN_ID or CONFIRM_PRODUCTION != "START_TRIGA2000_NSGAII":
    raise RuntimeError("Set a unique campaign ID and the production confirmation token.")
print("Command:", arguments)
```

## Execute

Dry-run checks inputs and runtime composition. Optimization starts or resumes
transport and may require substantial time. For long campaigns, use the terminal
launcher so process interruption and checkpoint status are easier to monitor.


```python
completed = subprocess.run(arguments, cwd=ROOT, check=False)
print("Exit code:", completed.returncode)
```
